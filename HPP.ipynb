### Cleaning Dataset 1:

```python
# Select and rename core columns
df1_clean = df1[['SalePrice', 'GrLivArea', 'FullBath', 'HalfBath', 'BedroomAbvGr', 'Neighborhood']].copy()
df1_clean.columns = ['Price', 'Area_SqFt', 'FullBath', 'HalfBath', 'Bedrooms', 'Location']

# Simple math for bathrooms: 1 full + 0.5 half
df1_clean['Bathrooms'] = df1_clean['FullBath'] + (0.5 * df1_clean['HalfBath'])
df1_clean = df1_clean.drop(['FullBath', 'HalfBath'], axis=1)

print(df1_clean.head(7))
```

```python
# Selecting columns
df2_clean = df2[['price', 'size', 'beds', 'baths', 'neighborhood']].copy()
df2_clean.columns = ['Price', 'Area_Raw', 'Bedrooms', 'Bathrooms', 'Location']

# Remove " sqft" and handle simple ranges by taking the first number
df2_clean['Area_SqFt'] = df2_clean['Area_Raw'].str.split('-').str[0]
df2_clean['Area_SqFt'] = df2_clean['Area_SqFt'].str.replace(' sqft', '').str.replace(',', '')
df2_clean['Area_SqFt'] = pd.to_numeric(df2_clean['Area_SqFt'], errors='coerce')

df2_clean = df2_clean.drop('Area_Raw', axis=1)
print(df2_clean.head(7))
```

```python
# Rename df3 columns to match df2_clean
df3_clean = df3[['price', 'bed', 'bath', 'city', 'house_size']].copy()
df3_clean.columns = ['Price', 'Bedrooms', 'Bathrooms', 'Location', 'Area_SqFt']
df3_clean
```

```python
df1_clean = df1_clean.dropna()
df2_clean = df2_clean.dropna()
df3_clean = df3_clean.dropna()
```

### Data Integration:

```python
# 1. Add a source tag to each cleaned dataset
df1_clean['Source'] = 'Ames_USA'
df2_clean['Source'] = '99acres_India'
df3_clean['Source'] = 'Bengaluru_India'

# 2. Vertically stack the datasets
# pandas will align the columns (Price, Area_SqFt, Bedrooms, Bathrooms, Location)
integrated_df = pd.concat([df1_clean, df2_clean, df3_clean], ignore_index=True)

# 3. Final cleaning on the combined dataset
# Ensure all numeric columns are floats
numeric_cols = ['Price', 'Area_SqFt', 'Bedrooms', 'Bathrooms']
integrated_df[numeric_cols] = integrated_df[numeric_cols].apply(pd.to_numeric, errors='coerce')

# 4.Creating a 'Price_Per_SqFt' column for better comparison
integrated_df['Price_Per_SqFt'] = integrated_df['Price'] / integrated_df['Area_SqFt']

# Saving the final integrated dataset
integrated_df.to_csv('integrated_property_data.csv', index=False)

print("Integration Complete!")
print(f"Total rows: {len(integrated_df)}")
print(integrated_df.head(7))
```
