```python
import numpy as np
import pandas as pd

n_samples = len(df)
```

```python
#1. Age_of_Property:
age = np.random.gamma(shape=2.0, scale=10.0, size=n_samples).astype(int)
age = np.clip(age,0,100)
```

```python
#2. Parking:
parking = np.random.choice([0,1,2,3,4], size=n_samples, p=[0.1,0.3,0.3,0.1,0.2])
```

```python
#3. Nearby_Schools:
schools = np.random.poisson(lam=3, size=n_samples)
schools = np.clip(schools, 0, 10)
```

```python
#4. Dist_to_city: 
dist = np.random.uniform(1.2,45,size=n_samples).round(2)
```

```python
#Adding the new features to the Dataframe:
df['Age_of_Property'] = age
df['Parking'] = parking
df['Nearby_Schools'] = schools
df['Dist_to_city'] = dist

print(df[['Age_of_Property','Parking','Nearby_Schools','Dist_to_city']].head(7) )
```

```python
df_synthetic = df
```

```python
df_synthetic.head()
```

```python
df_synthetic.shape
```

```python
#Converting to save memory:
df_synthetic['Age_of_Property'] = df_synthetic['Age_of_Property'].astype('int8')
df_synthetic['Parking'] = df_synthetic['Parking'].astype('int8')
df_synthetic['Nearby_Schools']  = df_synthetic['Nearby_Schools'].astype('int8')
df_synthetic['Dist_to_city'] = df_synthetic['Dist_to_city'].astype('float32')
```

```python
#Storing the dataset:
df_synthetic.to_csv('new_synthetic_dataset.csv', index=False)
```
